In this exercise you will apply the non-graphical EDA methods from the concept notebook to a new dataset: **Swiss public EV charging stations** (source: [Swiss Federal Office of Energy, OICP open data](https://opendata.swiss/en/dataset/ladestellen-fuer-elektroautos)).

## Tasks
1. Load the dataset and inspect its shape, data types, missing values and duplicates.
2. Compute mean, median and mode for `power_kw`.
3. Compute the 5th, 25th, 50th, 75th and 95th percentile for `power_kw`, `amperage` and `voltage`.
4. Compute variance and standard deviation for the numerical columns.
5. Compute skewness and kurtosis for `power_kw`. Interpret the results.
6. Build a contingency table of `power_type` vs. `plug_type`.
7. Compute a Pearson correlation matrix for the numerical columns and interpret the strongest correlation.

```python
# Libraries
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

### Task 1 — Load and inspect the dataset

```python
# Load the dataset
# YOUR CODE HERE
df = pd.read_csv('../data/ev_charging_stations.csv', sep=',')
print('Shape:', df.shape)
df.head(3)
```

```python
# Shape, data types, missing values, duplicates
# YOUR CODE HERE
print("Shape:",df.shape)
print("Datentypen: ",df.dtypes)
print("Missing Values: ",df.isnull().sum())
print("Duplicates: ",df.duplicated().sum())
```

```python
df
```

### Task 2 — Mean, median and mode of `power_kw`

```python
print("Mean: ",df['power_kw'].mean().round(2))
print("Median: ",df['power_kw'].median().round(2))
print("Mode: ",df['power_kw'].mode()[0].round(2))
```

### Task 3 — Quantiles for numerical columns

```python
# YOUR CODE HERE
df_quantile=df[['power_kw','amperage','voltage']].quantile([0.05,0.25,0.5,0.75,0.95]).round(2)
df_quantile
```

### Task 4 — Variance and standard deviation

```python
# YOUR CODE HERE
num_cols = ['power_kw', 'amperage', 'voltage']
print(df[num_cols].var().round(2))
print(df[num_cols].std().round(2))
```

### Task 5 — Skewness and kurtosis of `power_kw`

```python
# YOUR CODE HERE
print('Skewness:', round(df['power_kw'].skew(), 2))
print('Kurtosis:', round(df['power_kw'].kurt(), 2))
```

```python
# Interpretation (fill in the blanks):
# Skewness of power_kw: 3.0 → the distribution is strongly right-skewed because 3 is bigger than 0 and median is smaller than mean. few fast DC chargers with very high power
# Kurtosis of power_kw: 8.95 → the tails are heavier than a normal distribution
```

### Task 6 — Contingency table: `power_type` × `plug_type`

```python
contingency_table=pd.crosstab(df['power_type'], df['plug_type'], margins=True, margins_name='Total')
contingency_table
```

### Task 7 — Pearson correlation matrix

```python
korr_tab = df[num_cols].corr(method='pearson', numeric_only=True).round(3)
korr_tab
```

```python
# Interpretation:
# The strongest correlation is between amperage and power_kw (r = 0.997)
# This means: if the amperage is high, the power_kw is also high. it has a positive corelation, but it does not prove causality. 
# Stations with a higher current (amperage) deliver more power, because power = voltage × current (P = U × I).
```
